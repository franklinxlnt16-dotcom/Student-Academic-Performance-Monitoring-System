```python
# Import the necessary libraries
import pandas as pd
import numpy as np

# Load the file
df = pd.read_csv("/content/drive/MyDrive/Pairbud project/uci_scores_simple.csv")

# Inspect it first
print(df.shape)
print(df.head())
df.info()

# Check for missing values
print("\nMissing values:")
print(df.isnull().sum())

# Check for duplicate rows
print("\nDuplicate rows:", df.duplicated().sum())

# Each student should have exactly 3 periods
counts = df.groupby("student_id")["period"].count()
print("Number of students:", len(counts))
print("Students without exactly 3 periods:", (counts != 3).sum())

# Check that scores are between 0 and 100
print("\nLowest and highest scores:")
print(df[["Maths", "Portuguese"]].describe().loc[["min", "max"]])

# Check that absences are not negative
print("\nNegative absences:", (df["absences_maths"] < 0).sum() + (df["absences_portuguese"] < 0).sum())

# Check that the average column is correct
check = (df["Maths"] + df["Portuguese"]) / 2
print("Rows where average is wrong:", (abs(check - df["average"]) > 0.01).sum())

# Flag zero scores (they would be kept, not deleted)
df["has_zero"] = np.where((df["Maths"] == 0) | (df["Portuguese"] == 0), "Yes", "No")
print("\nRows with a zero score:", (df["has_zero"] == "Yes").sum())
print("Zeros in Maths:", (df["Maths"] == 0).sum())
print("Zeros in Portuguese:", (df["Portuguese"] == 0).sum())

# Validate and save
print(df.head())
df.to_csv("/content/drive/MyDrive/Pairbud project/cleaned_scores.csv", index=False)
print("\nSaved cleaned_scores.csv")
```
